```python
# Question 1 A - Import the dataset and display the first 10 rows

# import the library
import pandas as pd

# load the dataset
df=pd.read_csv('SensorLogs.csv')

# displaying first 10 rows
print(df.head(10))
```

```python
# Question 1 B - Handle missing PollutantLevel values by filling them with the median

# calculate the median of the PollutantLevel column
median_value = df['PollutantLevel'].median()
print(f'The median value of PollutantLevel is: {median_value}')

# Replace NaN values in PollutantLevel with the median value
df['PollutantLevel'].fillna(median_value, inplace=True)

# Verify
print ('THe missing values in pollutant level : ')
print (df['PollutantLevel'].isnull().sum())
```

```python
# Question 1 C - Create TotalImpact column (PollutantLevel x 1.25) and sort descending

# Create TotalImpact column
df['TotalImpact'] = df['PollutantLevel'] * 1.25 

# Sort descending by TotalImpact
df_sorted = df.sort_values(by='TotalImpact', ascending=False)


# Verify
print(df_sorted.head(10))
```

```python
# Question 1 D - Bin PollutantLevel into : Safe (0-30), Moderate (31-70), Hazardous (71+)

# Cutoff poitns : Anything up to 30, 30 to 70 and 70 above
bins = [-float('inf'), 30, 70, float('inf')]
labels = ['Safe', 'Moderate', 'Hazardous']

# create the new column
df['pollutant_category'] = pd.cut(df['PollutantLevel'], bins=bins, labels = labels)

# verifying the results
print(df[['PollutantLevel', 'pollutant_category']].head(10))
```

```python
# Question 1 E - Count occurences of each bin label

# count how many rows are in each category
category_counts = df['pollutant_category'].value_counts()
print(category_counts)

# count including the missing / null readings
count_with_nulls = df['pollutant_category'].value_counts(dropna = False)
print (f'With Nulls :\n{count_with_nulls}')
```

```python
# Question 1 F - Identify the ReadingID with the maximum pollutant level

# returns the index number
max_idx = df['PollutantLevel'].max()

# Retrieve the ReadingID at that index
reading_id = df[df['PollutantLevel'] == max_idx]
print (f'The max reading id is : \n{reading_id['ReadingID']}')
```
