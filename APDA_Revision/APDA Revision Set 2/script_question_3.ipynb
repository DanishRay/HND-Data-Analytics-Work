```python
# Question 3 A

import pandas as pd

# load the dataaset
df = pd.read_csv('OlympicMedals.csv')

# create a new DataFrame without duplicates
# using subset to specify columns and keep='last' to retain the final entry
df_cleaned = df.drop_duplicates(subset = ['Athlete', 'Event'], keep = 'last')
```

```python
# Question 3 B

import pandas as pd
import matplotlib.pyplot as plt

# aggregate medals by country
country_medals = df.groupby('Country')[['Gold', 'Silver', 'Bronze']].sum()

# calculate total and select top 10
country_medals['Total'] = country_medals.sum(axis=1)
top_10 = country_medals.sort_values(by = 'Total', ascending= False).head(10)

# create the plot with the colors given
colors = ['#FFD700', '#C0C0C0', '#CD7F32'] #Gold, Silver, Bronze
ax = top_10[['Gold', 'Silver', 'Bronze']].plot(
    kind = 'bar',
    stacked= True,
    color = colors,
    figsize= (10, 6)
)

plt.title ('Top 10 Countries by Medal Count')
plt.ylabel('Number of Medals')
plt.legend(title = 'Medal Type')
plt.show()
```
