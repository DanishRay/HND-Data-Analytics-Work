```python
# question 1 a

# import the dataset
import pandas as pd
df = pd.read_csv('SensorLogs.csv')

# displaying the last 5 rows
print(df.tail(5))
```

```python
# question 1 b

# calculate the mean of the PollutantLevel column
mean_value = df['PollutantLevel'].mean()

# verifying
print (f'The mean of Pollutant Level : {mean_value}')


# fill the missing values in the PollutantLevel column
df['PollutantLevel'] = df['PollutantLevel'].fillna(mean_value)

# check for the missing values
print(df.isnull().sum())
```

```python
# question 1 c

# the na=false argument ensures that if there are empty cells, they don't cause an error
filtered_df = df[df['AreaCode'].str.startswith('A', na=False)]

# verifying
print (filtered_df.head())
```

```python
# question 1 d

# calculate the average PollutantLevel
# grouped by AreaCode
avg_pollutant = df.groupby('AreaCode')['PollutantLevel'].mean()

print (f'Average Pollutant Level by Area Code : \n{avg_pollutant}')
```

```python
# question 1 e

# define the boundaries
# above 85 is 'Alert', otherwie its 'Normal'
bin_edges = [0, 85, float('inf')]

# define the names for the bin
bin_labels = ['Normal', 'Alert']

# perform the binning
df['Status'] = pd.cut(df['PollutantLevel'], bins = bin_edges, labels = bin_labels)

# previewing the new column
print (df[['PollutantLevel', 'Status']].head(10))
```

```python
# question 1 f

# returns the index number
min_idx = df['PollutantLevel'].min()

# retrieve the ReadingID at that index
reading_id = df[df['PollutantLevel'] == min_idx]
print (f'The lowest reading id is: \n{reading_id['ReadingID']}')
```
