```python
# question 3 a
import pandas as pd

# load the dataset
df = pd.read_csv('OlympicMedals.csv')

# using subset to specify columns and keep= 'first' to retain the first entry
df_cleaned = df.drop_duplicates(subset = ['Athlete', 'Event'], keep = 'first')
```

```python
# question 3 b
import matplotlib.pyplot as plt

# create the scatter plot
plt.scatter(df['Gold'], df['Silver'])

# add context, labels and title
plt.xlabel('Gold')
plt.ylabel('Silver')
plt.title('Gold vs Silver')

plt.show()
```

```python
# question 3 c
import seaborn as sns

# aggregate the data
df['Total'] = df[['Gold', 'Silver', 'Bronze']].sum(axis=1)

# group by athlete and sum the total column
athlete_total = df.groupby('Athlete')['Total'].sum()

# get the top 5 athletes by total medals
top_5 = athlete_total.sort_values(ascending=True).tail(5)

# creating the horizontal bar chart
ax = top_5.plot(
    kind = 'barh',
    color = 'skyblue',
    figsize=(10, 6)
)

plt.title('Total Medals for the Top 5 Athletes')
plt.xlabel('Total Number of Medals')
plt.ylabel('Athlete Name')
plt.show()
```

```python
# question 3 d

# group by event and calculate mean of the gold column
avg_gold = df.groupby('Event')['Gold'].mean()

print(avg_gold)


# find the name of the event with the highest average
best_event = avg_gold.idxmax()
highest_value = avg_gold.max()

print (f'The sport with the highest average Gold medals is {best_event} with {highest_value:.2f} medals')
```
