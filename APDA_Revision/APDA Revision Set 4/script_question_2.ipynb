```python
# question 2 a
import pandas as pd
df = pd.read_csv('PatientRecords.csv')

# data preparation feature selection
x = df.drop(columns = ['Diagnosis_Result'])

# create the target
y = df['Diagnosis_Result']


# perform data splitting
from sklearn.model_selection import train_test_split

# split the data
x_train, x_test, y_train, y_test = train_test_split(
    x,y,
    test_size = 0.2,
    random_state = 42,
    stratify = y
)


# apply standardization
from sklearn.preprocessing import StandardScaler

# initialize the scaler
scaler = StandardScaler()

# define for all numerical features
cols_to_scale = ['Age', 'Glucose_Level', 'BMI', 'BloodPressure']

# calculate the mean and deviation on the training data to all numerical features
x_train[cols_to_scale] = scaler.fit_transform (x_train[cols_to_scale])

# apply the same scaling to the test data
x_test[cols_to_scale] = scaler.transform(x_test[cols_to_scale])
```

```python
# question 2 b
# modelling

# importing KNN
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import recall_score, f1_score

# hyperparameter tuning
# initialize the base model
knn = KNeighborsClassifier()

# define the grid of parameters
param_grid = {
    'n_neighbors' : [3, 5, 7, 9, 11],
    'weights': ['uniform', 'distance']
}

# setup the grid search
grid_search = GridSearchCV(knn, param_grid, cv=5, scoring = 'f1')

# run the search on the training data
grid_search.fit(x_train, y_train)

print(f'Parameters found: {grid_search.param_grid}')


# use the model to predict the results for the test set
y_pred = grid_search.predict(x_test)

# calculate the metrics
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print (f'KNN Recall: {recall:.2f}')
print (f'KNN F1-Score: {f1:.2f}')
```

```python
# question 2 c
# decision tree classifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import recall_score, f1_score

# train and tune decision tree
# set max_depth to 5 and min_samples_split to 10
dt_model = DecisionTreeClassifier(max_depth = 5, min_samples_split = 10, random_state = 25)
dt_model.fit(x_train, y_train)

# evaluate
def evaluate_model(model, x_train, x_test, y_train, y_test):
    train_preds = model.predict(x_train)
    test_preds = model.predict(x_test)

    print(f'{type(model).__name__} Performance')
    print (f'Train - Recall: {recall_score(y_train, train_preds):.2f}, f1: {f1_score(y_train, train_preds):.2f}')
    print (f'Test - Recall: {recall_score(y_test, test_preds):.2f}, f1: {f1_score(y_test, test_preds):.2f}')

evaluate_model(dt_model, x_train, x_test, y_train, y_test)
```
