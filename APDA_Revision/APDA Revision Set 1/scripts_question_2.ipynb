```python
import pandas as pd
df = pd.read_csv('MembershipData.csv')
```

```python
# Question 2 A
# Data preparation

# Separate target from features
# Renewal_status is the target (y)
y = df['Renewal_Status']

# Everything else is the features (X)
x = df.drop(columns = ['Renewal_Status'])


# Normalize Monthly_Spend
from sklearn.preprocessing import MinMaxScaler
scalar = MinMaxScaler()

# Applying normalization to the Monthly_Spend column
x['Monthly_Spend'] = scalar.fit_transform(x[['Monthly_Spend']])


# Shuffle and Split with stratified sampling
from sklearn.model_selection import train_test_split

# Perform the split
x_train, x_test, y_train, y_test = train_test_split(
    x, y,
    test_size = 0.25,
    random_state = 25,
    stratify = y   
)
# Test Size (25%) - hold back 25% of the data for testing
# Sped (25) - random_state ensures that if you run the code again, will get the exact same split (reproductibiiity)
# Stratified Sampling - Crucial for classification.
# Ensures that the percentage of Renewed vs Not Renewed stays the same in both the training and testing seets
# Preventing a biased model


# Validation check
print (f'Training set size : {len(x_train)}')
print (f'Testing set size : {len(x_test)}')
```

```python
# Question 2 B
# Decision Tree - Training and Tuning
# Decision Tree can become overly complex if left unpruned
# Using hyperparameter like max_depth to stop it from growing too large
# and memorizing the training data

from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import recall_score, f1_score

# Train and tuen Decision Treee
# Set max_depth to 5 and min_samples_split to 10 
dt_model = DecisionTreeClassifier(max_depth = 5, min_samples_split = 10, random_state = 25)
dt_model.fit(x_train, y_train)

# Evalute
def evaluate_model(model, x_train, x_test, y_train, y_test):
    train_preds = model.predict(x_train)
    test_preds = model.predict(x_test)

    print(f'{type(model).__name__} Performance')
    print(f'Train - Recall : {recall_score(y_train, train_preds):.2f}, f1 : {f1_score(y_train, train_preds):.2f}')
    print(f'Test - Recall : {recall_score(y_test, test_preds):.2f}, f1 : {f1_score(y_test, test_preds):.2f}\n')

evaluate_model(dt_model, x_train, x_test, y_train, y_test)
```
