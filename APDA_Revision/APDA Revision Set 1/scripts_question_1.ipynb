```python
## Question 1 A
## Importing the library
import pandas as pd

## Load the dataset
df = pd.read_csv('SupplyChain.csv')

## To verify the import
print (df.head())
```

```python
## Question 1 B
## Calculate the mean of the ShippingCost Columns
mean_cost = df['ShippingCost'].mean()
print ('The average shipping cost is: ', mean_cost)

## To replace NaN values in ShippingCost with the mean_cost
df['ShippingCost'] = df['ShippingCost'].fillna(mean_cost)

## To verify the replacement
print ('The missing values in shipping cost : ')
print(df['ShippingCost'].isnull().sum())
```

```python
## Question 1 C
## NaN : A mathematical placeholder that tells Python, 'There is no data here'
## Empty String  :A piece of text that happens to have zero characters. Python treats this as valid data,
## which can break your calculations or charts

# Importing the library
import numpy as np

# Replace exact empty strings with NaM
df = df.replace ("", np.nan)

# Alternative way :- Using Regex
# r'^\s*$' is regest for : Start of string, any number of spaces, end of string
df = df.replace (r'^\s*$', np.nan, regex=True) 

# To verify ;- count missing value per column
print(df.isnull().sum())
```

```python
## Question 1 D
## Importing the numpy library
import numpy as np

## using a replace with a regular expression to find empty or whitespace-only strings
df = df.replace(r'^\s*$', np.nan, regex=True)
# Somtimes datasets contain empty spaces or actual empty strings instead of formal Null values
## Python treats these as text, which can cause errors in calculation
## To fix it, convert them to NaN, so that Pandas recognize them as missing

## Perform the binning
df ['ShipMode'] = pd.qcut(df['ShippingCost'],
q = 4, labels = ['Budget', 'Standard', 'Express', 'Overnight'])

## Verifying the distribution
print (df['ShipMode'].value_counts())

## Understanding the Results
# By looking at the output, it can be see how the cost were mapped
# Budget - Represent the cheapest 25% of shipping cost
# Standard - The next 25%
# Express - The 25% above median
# Overnight = The most expensive 25% of costs
```

```python
## Question 1 E
# Create a Mapping Dictionary
# Define the relationship between the old values (WarehouseiD) and the new values (Zone)
# in a Python dictionary.
# The keys are the existing IDs, and the values are the geographical zones

# The key is what we look for, the value is what we replace it with
zone_mapping = {
    'W01' : 'North',
    'W02' : 'North',
    'W03' : 'West',
    'W04' : 'West',
    'W05' : 'Central',
    'W06' : 'Central'
}

# Create the new column 'Zone'
df['Zone'] = df ['WarehouseID'].map(zone_mapping)

# Check the first few rows to see the pairing
print (df[['WarehouseID', 'Zone']].head())

# Checking if Zones are empty
print(df['Zone'].isnull().sum())

## Mapping is for
# Speed - Dictionary lookups in Python (O(1) complexity)
# Readability - It is much easier to read a dictionary that a long list of if-else statments
# Maintenance - If a new warehouse (W07) is added of the North zone, only have to update one line in the dictionary

```

```python
# Question 1 F

# Calculate Total Shipping Cost for each Zone
# Using the .groupby() method.
# It splits the data into groups based on the Zone and then aaplied the sum()
# function to the ShippingCost column

# Group by Zone and calculate the sum of ShippingCost
zone_shipping = df.groupby('Zone')['ShippingCost'].sum()

print (zone_shipping)

#----------------------------------------------------------------------#
# Find Month with highest / lowest order volume
# Need to ensure the OrderDate column is recognized as date
# not just text

# Convert the string date to a proper datetime object
df['OrderDate'] = pd.to_datetime(df['OrderDate'])

# Create a month column
df['Month'] = df['OrderDate'].dt.month_name()

# Count orders per month
month_counts = df['Month'].value_counts()

# Find the Max and MIn
# idmax() gives the index (month name) with the highest value
highest_month = month_counts.idxmax()
lowest_month = month_counts.idxmin()

print (f'Highest Volume : {highest_month}')
print (f'Lowest Volume : {lowest_month}')
```
