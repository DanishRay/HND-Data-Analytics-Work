```python
# Question A

# need to import the pandas library to read the csv file
import pandas as pd

# Load the dataset
df = pd.read_csv('TechSales.csv')

# Selecting the target column 'Profit'
profit_column = df['Profit']

# Calculate descriptive statistics
min_profit = profit_column.min()
max_profit = profit_column.max()
mean_profit = profit_column.mean()

# Print the results
print(f"Minimum Profit: {min_profit}")
print(f"Maximum Profit: {max_profit}")
print(f"Mean Profit: {mean_profit}")
```

```python
# Question B

# need to import pandas and plotting library which is matplotlib
import pandas as pd
import matplotlib.pyplot as plt

# load the dataset
df = pd.read_csv('TechSales.csv')

# Create teh scatter plot
plt.scatter(df['Revenue'], df['Profit'])

# Add the context, labels and title
plt.xlabel('Revenue')
plt.ylabel('Profit')    
plt.title('Revenue vs Profit')

# show the plot
plt.show()
```

```python
# Question C

# Importing the necessary libraries
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Aggregate the data
# calculate the average revenue for each product per region
df_grouped = df.groupby(['Product', 'Region'])['Revenue'].mean().reset_index()

# Define custom hex colors
colors = ['#FF5733', '#33FF57', '#3357FF', '#F333FF', '#33FFF5']

# Create the Grouped Bar chart
plt.figure(figsize=(12, 6))
sns.barplot(x='Product', y='Revenue', hue='Region', data=df_grouped, palette=colors)

# Add a title and move the legend so it doesn't overlap with the bars
plt.title('Average Revenue by Product and Region')
plt.ylabel('Average Revenue ($)')
plt.legend(title='Region', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.show()
```
