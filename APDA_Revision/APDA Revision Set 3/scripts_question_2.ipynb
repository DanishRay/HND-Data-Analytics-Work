```python
# question 2 a Data Preparation

# importing library pandas
import pandas as pd

# load the dataset
df = pd.read_csv('MembershipData.csv')


# separate features from the target variabale
# create the features (x) by dropping the target column
x = df.drop(columns = ['Renewal_Status'])

# create the target (y)
y = df['Renewal_Status']


# perform a data split (train-test split)
from sklearn.model_selection import train_test_split

# split the data
x_train, x_test, y_train, y_test = train_test_split(
    x,y,
    test_size = 0.2,
    random_state = 42,
    stratify = y
)


# apply standarization (z-score scaling)
from sklearn.preprocessing import StandardScaler

# initialize the scaler
scaler = StandardScaler()

# calculate the mean and deviation on the training data and transform it
x_train['Tenure'] = scaler.fit_transform(x_train[['Tenure']])

# apply the same scaling to the test data 
x_test['Tenure'] = scaler.transform(x_test[['Tenure']])
```

```python
# question 1 b

# train a random forest classifier
from sklearn.ensemble import RandomForestClassifier

# initialize the model with tuned hyperparameter
rf_model = RandomForestClassifier(n_estimators = 100, max_depth = 5, random_state = 42)

# train the model using the training data
rf_model.fit(x_train, y_train)


# evaluate the Random Forest
from sklearn.metrics import precision_score, f1_score

# get predictions for both sets
rf_train_preds = rf_model.predict(x_train)
rf_test_preds = rf_model.predict(x_test)

# calculate scores
print (f'RF Train Precision: {precision_score(y_train, rf_train_preds):.2f}')
print (f'RF Test Precision: {precision_score(y_test, rf_test_preds):.2f}')
print (f'RF Test F1-Score: {f1_score(y_train, rf_train_preds):.2f}')


# train a logistic regression model
from sklearn.linear_model import LogisticRegression

# initialize and train the model
log_model = LogisticRegression(C=0.1, solver='liblinear', random_state = 42)
log_model.fit(x_train, y_train)


# evaluate logistic regression
# get predictions for the test set
log_test_preds = log_model.predict(x_test)

# calculate scores
print (f'Logistic Regression Test Precision: {precision_score(y_test, log_test_preds):.2f}')
print (f'Logistic Regression F1-Score: {f1_score(y_test, log_test_preds):.2f}')
```

### Model Comparison (Minimizing False Positives)

A **False Positive** occurs when the model predicts a customer **will renew** (Positive), but they actually **churn/leave** (False). This is a critical error for business because the company might ignore a customer who was actually about to leave, missing the chance to offer them a retention discount or incentive.
