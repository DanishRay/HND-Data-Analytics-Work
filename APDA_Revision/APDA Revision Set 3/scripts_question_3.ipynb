```python
# question 3 A

# load the dataset
import pandas as pd
df = pd.read_csv('TechSales.csv')


# group by Product and calculate sum and mean for Revenue
summary_table = df.groupby('Product')['Revenue'].agg(['sum', 'mean'])

# rename the columns for better readability
summary_table = summary_table.rename(columns = {
    'sum' : 'Total Revenue',
    'mean' : 'Average Revenue'
})

# display the table
print (summary_table)
```

```python
# question 3 b

# import visualization libraries
import matplotlib.pyplot as plt
import seaborn as sns

# create the boxplot
# set the size of the plot
plt.figure(figsize = (10, 6))

sns.boxplot(x = 'Region', y = 'Profit', data=df)

# add labels and title
plt.title('Distribution of Profit by Region')
plt.xlabel('Region')
plt.ylabel('Profit (USD)')

plt.show()
```

```python
# question 3 c

import matplotlib.pyplot as plt
import seaborn as sns

# set the size of the chart
plt.figure(figsize = (10, 7))

# creating the horizontal bar chart
sns.barplot(
    x = 'Profit',
    y = 'Region',
    hue = 'Product',
    data = df,
    palette = 'viridis',
    estimator = 'sum',
    errorbar = None # this removes the small lines for a cleaner look
)


# add title and labels
plt.title('Total Profit by Region and Product')
plt.xlabel('Total Profit(USD)')
plt.ylabel('Region')

# move the legend outside if it blocks the bars
plt.legend(title = 'Product', bbox_to_anchor = (1.05, 1), loc = 'upper left')

# adjusting the layout so nothing is cut off
plt.tight_layout()

# display
plt.show()
```
