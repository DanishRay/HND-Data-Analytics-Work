```python
# question 1 a

# import library pandas 
import pandas as pd

# load the dataset into a dataframe
df = pd.read_csv('SupplyChain.csv')

# veryifying
print (df.head()) #display the rows

print (df.info()) # display dataset info
```

```python
# question 1 b

# calculate the median of the ShippingCost column
median_value = df['ShippingCost'].median()

# verifying
print (f'The median shipping cost is: {median_value}')

# fill the missing values in the ShippingCost column
df['ShippingCost'] = df['ShippingCost'].fillna(median_value)

# check  for remaining missing values
print(df.isnull().sum())
```

```python
# question 1 c

# convert all values in the Category column to uppercase .str.upper()
df['Category'] = df['Category'].str.upper()

# verify the change
print(df['Category'].unique())
```

```python
# question 1 d

# define the boundaries for the bins
# 0 to 35 is bin 1, 35 to 75 is bin 2, 75 and above is bin 3
bin_edges = [0, 35, 75, float('inf')]

# defining the names for the bins
bin_labels = ['Economy', 'Standard', 'Premium']

# perform the binning
df['ServiceLevel'] = pd.cut(df['ShippingCost'], bins = bin_edges, labels = bin_labels)

# preview the new column alongside the original cost
print(df[['ShippingCost', 'ServiceLevel']].head(10))

# check the count of each category
print(df['ServiceLevel'].value_counts())
```

```python
# question 1 e

# define the mapping logic
warehouse_map = {
    'W01':'Urban',
    'W03':'Urban',
    'W05':'Urban',
    'W02':'Rural',
    'W04':'Rural',
    'W06':'Rural'
}

# create the new column using the map
df['LocationType'] = df['WarehouseID'].map(warehouse_map)

# check the first few rows of the mapping
print(df[['WarehouseID', 'LocationType']].head())

# ensuring there are no unmapped values
print(df['LocationType'].unique())
```

```python
# question 1 f

# calculate the average ShippinCost for each category
# group the data by Category and calculate the mean of ShippinCost
avg_shipping = df.groupby('Category')['ShippingCost'].mean()

print (f'Average Shipping Cost by category: \n{avg_shipping}')


# identify the WarehouseID with the highest orders
# count how many times each WarehouseID appears
warehouse_counts = df['WarehouseID'].value_counts()

# identify the top one
top_warehoues = warehouse_counts.idxmax()
order_count = warehouse_counts.max()
print (f'The warehouse with the most orders is {top_warehoues} with {order_count} orders')
```
